# 1. About these notebooks

## 1.1 About

These notebooks follow the online [QuantEcon](https://quantecon.org/) lecture series, available for free at the linked resource. In this first series of notebooks I follow along with the intruductory QuantEcon lecture series, [A First Course in Quantitative Economics with Python](https://intro.quantecon.org/intro.html). This first series serves as a foundation for quantitative economic analysis and modeling, focusing on visualization and simulation with code rather than mathematical details.

## 1.2 Level

The lecture series itself is aimed at second and third year undergraduate series who are familiar with calculaus and linear algebra. The content varies from the introductory level (some of which is even accessible to high schoolers) to more intermediate topics. The easier material is generally found earlier on in the series, while more intermediate material requiring knowledge of linear algebra is found later on.

As the author of these particular notebooks, I hold a PhD in astrophysics and consider myself to have a robust mathematical foundation for these topics. However, despite having previously worked in marketing data science roles, I consider myself to be quite a novice when it comes to economic topics, no more versed than the average, technically literate American adult. These notebooks record my forray into a building a more structured understanding of economics and economic modeling through QuantEcon's self study program. I consider the authors of this program very generous for making these resources freely available.
